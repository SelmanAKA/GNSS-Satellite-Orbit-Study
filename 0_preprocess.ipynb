### **read_data**

```python
 # 2 14  8 28 10  0  0.0 5.151415243745D-04 2.160049916711D-12 0.000000000000D+00
def read_data(data_file, first_row, number_of_lines):
    data = []
    data_file.seek(0)

    for row in data_file:
        if row.split() == first_row.split():
            data.append(row)

            for i in range(number_of_lines - 1):
                data.append(next(data_file))

            break

    return data
```

```python
G02_raw = read_data(nav_file," 2 14  8 28 10  0  0.0 5.151415243745D-04 2.160049916711D-12 0.000000000000D+00",8)
G02_raw
```

### **split_data**

```python
def split_data(data):
    x = []
    for i,line in enumerate(data):
        line = line[23:] if i == 0 else line[3:]
        for j in range(0, len(line), 19):
            value = line[j:j+19].strip()

            if value:
                x.append(float(value.replace('D', 'E')))

    return x
```

```python
G02_data = split_data(G02_raw)
```
